```python
import random

ganadas = 0
perdidas = 0

while True:
    secreto = random.randint(1, 100)
    historial = []
    acerto = False
    print("\nAdivina el número entre 1 y 100 (máximo 7 intentos)")
    for intento in range(1, 8):
        while True:
            try:
                numero = int(input(f"Intento {intento}: "))
                if numero >= 1 and numero <= 100:
                    break
                print("Debe estar entre 1 y 100")
            except ValueError:
                print("Escriba un número entero")
        historial.append(numero)
        if numero == secreto:
            print("¡Adivinaste!")
            acerto = True
            break
        elif numero < secreto:
            print("Mayor")
        else:
            print("Menor")

    if acerto:
        ganadas = ganadas + 1
    else:
        perdidas = perdidas + 1
        print("El número era:", secreto)
    print("Historial:", historial)
    print("Ganadas:", ganadas, "Perdidas:", perdidas)
    otra = input("¿Jugar otra vez? (s/n): ").strip().lower()
    if otra != "s":
        break
```

```python
import random

def contar_vecinos(tablero, fila, columna):
    total = 0
    for cambio_fila in range(-1, 2):
        for cambio_columna in range(-1, 2):
            if cambio_fila == 0 and cambio_columna == 0:
                continue
            nueva_fila = fila + cambio_fila
            nueva_columna = columna + cambio_columna
            if 0 <= nueva_fila < 10 and 0 <= nueva_columna < 10:
                total = total + tablero[nueva_fila][nueva_columna]
    return total

def siguiente_generacion(tablero):
    nuevo = []
    for fila in range(10):
        nueva_fila = []
        for columna in range(10):
            vecinos = contar_vecinos(tablero, fila, columna)
            if tablero[fila][columna] == 1:
                if vecinos == 2 or vecinos == 3:
                    nueva_fila.append(1)
                else:
                    nueva_fila.append(0)
            else:
                if vecinos == 3:
                    nueva_fila.append(1)
                else:
                    nueva_fila.append(0)
        nuevo.append(nueva_fila)
    return nuevo

def mostrar_tablero(tablero):
    for fila in tablero:
        texto = ""
        for celda in fila:
            if celda == 1:
                texto = texto + "■ "
            else:
                texto = texto + "· "
        print(texto)

tablero = []
for fila in range(10):
    nueva_fila = []
    for columna in range(10):
        nueva_fila.append(random.randint(0, 1))
    tablero.append(nueva_fila)

for generacion in range(1, 6):
    print("\nGeneración", generacion)
    mostrar_tablero(tablero)
    tablero = siguiente_generacion(tablero)
```
