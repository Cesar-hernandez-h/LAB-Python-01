```python
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ").strip()
    if nombre.lower() == "fin":
        break
    while True:
        try:
            precio = float(input("Precio: "))
            if precio < 0:
                print("El precio no puede ser negativo")
                continue
        except ValueError:
            print("Error: escriba un precio numérico")
        else:
            print("Precio validado")
            break
        finally:
            print("Fin del intento de lectura del precio")
    while True:
        try:
            cantidad = int(input("Cantidad: "))
            if cantidad < 0:
                print("La cantidad no puede ser negativa")
                continue
        except ValueError:
            print("Error: escriba una cantidad entera")
        else:
            print("Cantidad validada")
            break
        finally:
            print("Fin del intento de lectura de la cantidad")
    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

print("Inventario validado:", productos)
```

```python
import csv

def guardar_csv(inventario, nombre_archivo):
    with open(nombre_archivo, "w", encoding="utf-8", newline="") as archivo:
        escritor = csv.writer(archivo)
        escritor.writerow(["nombre", "precio", "cantidad"])
        for producto in inventario:
            escritor.writerow([producto["nombre"], producto["precio"], producto["cantidad"]])

def leer_csv(nombre_archivo):
    inventario = []
    try:
        with open(nombre_archivo, "r", encoding="utf-8", newline="") as archivo:
            lector = csv.DictReader(archivo)
            for fila in lector:
                producto = {
                    "nombre": fila["nombre"],
                    "precio": float(fila["precio"]),
                    "cantidad": int(fila["cantidad"])
                }
                inventario.append(producto)
    except FileNotFoundError:
        print("El CSV no existe; se empieza con inventario vacío")
    return inventario

guardar_csv(productos, "inventario.csv")
productos_desde_csv = leer_csv("inventario.csv")
print("Inventario recuperado del CSV:", productos_desde_csv)
```

```python
import json

def guardar_json(inventario, nombre_archivo):
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        json.dump(inventario, archivo, ensure_ascii=False, indent=4)

def leer_json(nombre_archivo):
    try:
        with open(nombre_archivo, "r", encoding="utf-8") as archivo:
            inventario = json.load(archivo)
    except FileNotFoundError:
        print("El JSON no existe; se empieza con inventario vacío")
        inventario = []
    return inventario

guardar_json(productos, "inventario.json")
productos_desde_json = leer_json("inventario.json")
print("Inventario recuperado del JSON:", productos_desde_json)
```
